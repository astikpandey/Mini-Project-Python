```python

import numpy as np
import pandas as pd
import os

# Create or load contact file
file = "contacts.csv"

if os.path.exists(file):
    contacts = pd.read_csv(file, dtype=str)
else:
    contacts = pd.DataFrame(
        columns=["Name", "Phone", "Email"]
    )


# 1. Add Contact
def add_contact():
    global contacts

    name = input("Enter contact name: ").strip()
    phone = input("Enter phone number: ").strip()
    email = input("Enter email: ").strip()

    if name == "" or phone == "":
        print("Name and phone are required!")
        return

    # NumPy array
    phone_array = np.array(contacts["Phone"])

    if phone in phone_array:
        print("Phone number already exists!")
        return

    new_contact = pd.DataFrame({
        "Name": [name],
        "Phone": [phone],
        "Email": [email]
    })

    contacts = pd.concat(
        [contacts, new_contact],
        ignore_index=True
    )

    contacts.to_csv(file, index=False)

    print("Contact added successfully!")


# 2. View Contacts
def view_contacts():

    if contacts.empty:
        print("No contacts found!")

    else:
        print("\n------ Contact List ------")
        print(contacts.to_string(index=False))


# 3. Search Contact
def search_contact():

    name = input("Enter name to search: ").strip()

    result = contacts[
        contacts["Name"].str.contains(
            name, case=False, na=False, regex=False
        )
    ]

    if result.empty:
        print("Contact not found!")

    else:
        print("\nContact Found:")
        print(result.to_string(index=False))


# 4. Update Contact
def update_contact():
    global contacts

    name = input("Enter name to update: ").strip()

    result = contacts[
        contacts["Name"].str.lower() == name.lower()
    ]

    if result.empty:
        print("Contact not found!")
        return

    index = result.index[0]

    new_phone = input("Enter new phone (Enter to skip): ").strip()
    new_email = input("Enter new email (Enter to skip): ").strip()

    if new_phone != "":
        phone_array = np.array(contacts["Phone"])

        if new_phone in phone_array and new_phone != contacts.at[index, "Phone"]:
            print("Phone number already exists!")
            return

        contacts.at[index, "Phone"] = new_phone

    if new_email != "":
        contacts.at[index, "Email"] = new_email

    contacts.to_csv(file, index=False)

    print("Contact updated successfully!")


# 5. Delete Contact
def delete_contact():
    global contacts

    name = input("Enter name to delete: ").strip()

    result = contacts[
        contacts["Name"].str.lower() == name.lower()
    ]

    if result.empty:
        print("Contact not found!")

    else:
        contacts = contacts.drop(result.index)
        contacts = contacts.reset_index(drop=True)

        contacts.to_csv(file, index=False)

        print("Contact deleted successfully!")


# Main Program
while True:

    print("\n====== CONTACT MANAGER ======")
    print("1. Add Contact")
    print("2. View Contacts")
    print("3. Search Contact")
    print("4. Update Contact")
    print("5. Delete Contact")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ").strip()

    if choice == "1":
        add_contact()

    elif choice == "2":
        view_contacts()

    elif choice == "3":
        search_contact()

    elif choice == "4":
        update_contact()

    elif choice == "5":
        delete_contact()

    elif choice == "6":
        print("Thank you for using Contact Manager!")
        break

    else:
        print("Invalid choice! Please enter 1-6.")
```
